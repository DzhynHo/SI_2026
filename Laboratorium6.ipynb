```python

import os
import requests
from bs4 import BeautifulSoup
from IPython.display import Markdown, display
from groq import Groq
```

```python

client = Groq(api_key="YOUR_API_KEY")

response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role": "user", "content": "Tell me something about transformer technology"}]
)
print(response.choices[0].message.content)
```

```python

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}

class Website:
    def __init__(self, url):
        self.url = url
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, 'html.parser')
        self.title = soup.title.string if soup.title else "No title found"
        for irrelevant in soup.body(["script", "style", "img", "input", "link"]):
            irrelevant.decompose()
        self.text = soup.body.get_text(separator="\n", strip=True)
```

```python

web = Website("https://pl.wikipedia.org/wiki/Du%C5%BCy_model_j%C4%99zykowy")
print("Title:", web.title)
print("Text:", web.text[:500])
```

```python

system_prompt = (
    "You are an assistant that analyzes the contents of a website "
    "and provides a short summary, ignoring text that might be navigation related. "
    "Respond in markdown in Polish."
)
```

```python

def user_prompt_for(website):
    user_prompt = f"You are looking at a website titled {website.title}"
    user_prompt += "\nThe contents of this website is as follows; please provide a short summary in markdown. "
    user_prompt += "If it includes news or announcements, then summarize these too.\n\n"
    user_prompt += website.text
    return user_prompt
```

```python

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_for(website)}
    ]

web = Website("https://wp.pl")
messages_for(web)
```

```python

MODEL = "llama-3.3-70b-versatile"

def summarize(url):
    website = Website(url)
    response = client.chat.completions.create(
        model=MODEL,
        messages=messages_for(website)
    )
    return response.choices[0].message.content
```

```python

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))

display_summary("https://wp.pl")
```

# Zadanie – porównanie modeli


```python

modele = [
    "llama-3.3-70b-versatile",
    "llama-3.1-8b-instant",
    "meta-llama/llama-4-scout-17b-16e-instruct"
]
url = "https://wp.pl"
website = Website(url)
website.text = website.text[:3000]

for model in modele:
    print(f"\n{'='*50}\nModel: {model}\n{'='*50}")
    response = client.chat.completions.create(
        model=model,
        messages=messages_for(website)
    )
    display(Markdown(response.choices[0].message.content))
```

# Przetestowano trzy modele na stronie wp.pl:

**llama-3.3-70b-versatile** – najbardziej szczegółowe i płynne podsumowanie, pełne zdania, bogaty opis wszystkich sekcji strony.
**llama-3.1-8b-instant** – podsumowanie skrótowe, mniej szczegółów, niektóre informacje pominięte. Model szybszy, ale słabszy jakościowo.
**llama-4-scout-17b** – dobra struktura odpowiedzi z czytelnymi nagłówkami, jakość zbliżona do modelu 70b.

**Wniosek**: Większe modele generują dokładniejsze i bardziej szczegółowe podsumowania. Model 8b jest szybszy, ale widoczna jest różnica w jakości. Prompt systemowy z instrukcją „Respond in markdown in Polish" skutecznie wymuszał odpowiedzi w języku polskim we wszystkich testowanych modelach.
